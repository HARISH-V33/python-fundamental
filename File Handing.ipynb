# 01. Text Operations & Context Managers

### Step 1: Resource Management with `with`
The **`with`** statement is a Python context manager that guarantees proper file descriptor cleanup. It automatically invokes `__exit__()` and closes the file handle even if unhandled exceptions occur mid-read/write.

### Step 2: Access Modes
* `'w'`: Overwrites existing file content or creates a new file.
* `'a'`: Appends new records to the end of an existing file.
* `'r'`: Read-only mode (raises `FileNotFoundError` if file is missing).

```python
text_file = "notes.txt"

# 1. Writing content using 'w' mode
with open(text_file, "w", encoding="utf-8") as f:
    f.write("Line 1: Context Managers ensure zero resource leaks.\n")
    f.write("Line 2: Modern Python utilizes explicit UTF-8 encoding.\n")

# 2. Appending content using 'a' mode
with open(text_file, "a", encoding="utf-8") as f:
    f.write("Line 3: Appended entry using append mode.\n")

# 3. Reading entire file into a single string (`read()`)
print("--- Full Reading Output (`read()`) ---")
with open(text_file, "r", encoding="utf-8") as f:
    print(f.read())

# 4. Reading file into a list of line strings (`readlines()`)
print("--- Line-By-Line Output (`readlines()`) ---")
with open(text_file, "r", encoding="utf-8") as f:
    lines = f.readlines()
    for idx, line in enumerate(lines, start=1):
        print(f"[{idx}] {line.strip()}")
```

###  Code Explanation 
* **`encoding="utf-8"`**: Explicitly prevents platform-dependent encoding mismatches (Windows CP1252 vs Linux UTF-8).
* **`read()` vs `readlines()`**: `read()` loads the entire file stream into memory as one continuous string. `readlines()` parses the file into an ordered list of strings broken by `\n`.
* **`line.strip()`**: Removes trailing newline characters (`\n`) for clean console display.

# 02. Structured Serialization: CSV, JSON & Pickle

### Step 1: Comparative Overview
| Format | Module | Type | Primary Use Case |
|---|---|---|---|
| **CSV** | `csv` | Text (Tabular) | Spreadsheets, raw dataset exports |
| **JSON** | `json` | Text (Structured) | Web API payloads, application configs |
| **Pickle** | `pickle` | Binary (`bytes`) | Native Python object state persistence |

```python
import csv
import json
import pickle

# 1. CSV Processing (DictWriter / DictReader)
csv_file = "users.csv"
users_data = [
    {"id": 101, "name": "Vetri", "role": "Developer"},
    {"id": 102, "name": "Sadhana", "role": "Data Analyst"},
]

with open(csv_file, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["id", "name", "role"])
    writer.writeheader()
    writer.writerows(users_data)

print("--- Reading CSV Data ---")
with open(csv_file, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"User ID: {row['id']} | Name: {row['name']} | Role: {row['role']}")

# 2. JSON Processing (dump / load)
json_file = "config.json"
app_config = {
    "app_name": "Portal",
    "version": 2.1,
    "modules": ["logging", "file_handling"],
}

with open(json_file, "w", encoding="utf-8") as f:
    json.dump(app_config, f, indent=4)

with open(json_file, "r", encoding="utf-8") as f:
    loaded_config = json.load(f)

print("\n--- Loaded JSON Data ---")
print("App Name:", loaded_config["app_name"])

# 3. Pickle Binary Serialization (wb / rb)
pickle_file = "state.pkl"
native_object = {"weights": [0.1, 0.5, 0.9], "status": "Ready"}

with open(pickle_file, "wb") as f:
    pickle.dump(native_object, f)

with open(pickle_file, "rb") as f:
    restored_object = pickle.load(f)

print("\n--- Restored Pickle Object ---")
print(restored_object)
```

###  Code Explanation
* **`newline=""` in CSV open**: Prevents `csv.writer` from inserting blank line artifacts on Windows OS.
* **`DictWriter` & `DictReader`**: Automatically maps rows directly to Python dictionaries using header keys.
* **`json.dump(..., indent=4)`**: Formats JSON output with readable 4-space indentation hierarchy.
* **`wb` / `rb` modes**: Binary read/write modes required by `pickle` to handle raw Byte Streams.

# 03. File System Utilities & Operations (`pathlib`, `os`, `shutil`)

### Step 1: Tool Roles
* **`pathlib.Path`**: Object-oriented, cross-platform path manipulation engine.
* **`shutil`**: High-level utility for copying, moving, and recursively removing directory trees.
* **Validation**: Defensive programming checks (`exists()`, `is_file()`, `is_dir()`) before executing disk operations.

```python
import os
from pathlib import Path
import shutil

# 1. Creating workspaces with pathlib
base_dir = Path.cwd() / "demo_file_workspace"
base_dir.mkdir(exist_ok=True)

sample_file = base_dir / "sample.txt"
sample_file.write_text("Pathlib write_text execution", encoding="utf-8")

# 2. File Validation
print("--- Defensive File Checks ---")
print(f"Path Exists?: {sample_file.exists()}")
print(f"Is File?: {sample_file.is_file()}")
print(f"Is Directory?: {sample_file.is_dir()}")
print(f"File Size: {sample_file.stat().st_size} bytes")

# 3. Directory & Copy Operations
backup_dir = base_dir / "backup"
backup_dir.mkdir(exist_ok=True)

copied_file = backup_dir / "sample_copy.txt"
shutil.copy(sample_file, copied_file)
print(f"\nCopied file successfully to: {copied_file.name}")

# 4. Pattern Search with glob
print("\n--- Searching Workspace with rglob ---")
for text_path in base_dir.rglob("*.txt"):
    print(f"Discovered: {text_path.relative_to(base_dir)}")

# 5. Cleanup
shutil.rmtree(base_dir)
print(f"\nDirectory '{base_dir.name}' cleaned up successfully.")
```

###  Code Explanation
* **`base_dir = Path.cwd() / "demo_file_workspace"`**: Uses overloaded `/` operators in `pathlib` to construct cross-platform path objects smoothly.
* **`sample_file.write_text()`**: Direct shorthand helper in `pathlib` that handles opening, writing, and closing files without writing explicit `with open()` blocks.
* **`base_dir.rglob("*.txt")`**: Executes recursive globbing to discover matching text files across all nested subdirectories.
* **`shutil.rmtree()`**: Deletes a directory and all nested contents recursively (unlike `os.rmdir()` which fails if directories are non-empty).
