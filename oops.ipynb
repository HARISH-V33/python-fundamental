# 01. Foundations of OOP (Class, Object, `__init__`, Variables, Methods)

### Step 1: Understand the Concept
Object-Oriented Programming (OOP) structures code by grouping related data and behaviors into blueprints called **Classes**, from which individual **Objects** are instantiated.

* **Class**: A user-defined prototype or blueprint defining attributes and methods.
* **Object**: A concrete instance created from a class blueprint.
* **`__init__` (Constructor)**: A special initialization method automatically executed upon object creation.
* **`self`**: A mandatory reference parameter representing the specific instance executing the method.
* **Instance Variables**: Data attributes unique to each instance of a class.
* **Class Variables**: Shared attributes shared across all instances of a class.
* **Instance Methods**: Functions defined inside a class that operate on individual instance attributes via `self`.

### Step 2: Real-World Analogy
* **Class**: The architectural blueprint of a house.
* **Object**: The physical house constructed from the blueprint.
* **`__init__`**: The initial setup crew painting walls and setting up room layouts during construction.
* **Class Variable**: The community lawn shared equally by all houses in the neighborhood.
* **Instance Variable**: The specific interior decorations and furniture inside a single house.

### Step 3: Implementation & Explanation

```python
class Car:
    # Class Variable (shared across all instances)
    wheels = 4

    def __init__(self, brand: str, model: str, color: str):
        # Instance Variables (unique to each instance)
        self.brand = brand
        self.model = model
        self.color = color

    def display_info(self) -> str:
        """Instance method operating on instance attributes."""
        return f"{self.color} {self.brand} {self.model} with {Car.wheels} wheels"

    def drive(self, speed: int) -> None:
        """Instance method demonstrating state behavior."""
        print(f"The {self.brand} {self.model} is driving at {speed} km/h.")


# Instantiating Objects
car1 = Car("Swift", "VXi", "Red")
car2 = Car("BMW", "M3", "Black")

# Accessing Instance Variables & Class Variables
print("Car 1 Info:", car1.display_info())
print("Car 2 Info:", car2.display_info())

# Dynamic updates
car1.drive(80)
car2.drive(120)

# Verifying Class vs Instance Variables
print("\nClass Attribute Access:")
print("Shared Wheel Count via Class:", Car.wheels)
print("Shared Wheel Count via Instance 1:", car1.wheels)
```

**Code Explanation:**
* `class Car:` defines the blueprint structure.
* `wheels = 4` creates a class variable shared by all `Car` objects.
* `def __init__(self, brand, model, color):` initializes instance attributes when `Car()` is invoked.
* `self.brand`, `self.model`, `self.color` bind passed values to the specific object memory reference.
* `car1 = Car(...)` instantiates an object and triggers `__init__` under the hood.
* `car1.drive(80)` calls an instance method passing `car1` implicitly as `self`.

# 02. Class Methods, Static Methods, and Properties (`@property`)

### Step 1: Understand the Concept
Python provides specialized decorators to alter method behavior within classes:
* **`@classmethod`**: Binds the method to the class rather than the instance. Receives `cls` as the implicit first argument. Frequently used for alternative constructors.
* **`@staticmethod`**: Isolated utility functions inside a class namespace that do not receive `self` or `cls`.
* **`@property`**: Enables pythonic getter, setter, and deleter patterns for controlled attribute access and validation.

### Step 2: Real-World Analogy
* **Class Method**: A factory manager adjusting the global production count across all assembly lines.
* **Static Method**: A calculator mounted on the wall of the factory—useful for basic math, but doesn't know anything about factory operations.
* **Property Decorator**: A security guard at the gate checking if incoming materials meet safety standards before allowing them inside.

### Step 3: Implementation & Explanation

```python
class Employee:
    # Class Variable
    company_name = "Stackly"

    def __init__(self, name: str, salary: float):
        self.name = name
        self._salary = salary  # Encapsulated attribute

    # 1. Class Method (Alternative Constructor)
    @classmethod
    def from_string(cls, employee_str: str):
        """Creates an Employee object from a formatted string 'Name-Salary'."""
        name, salary = employee_str.split("-")
        return cls(name, float(salary))

    # 2. Static Method (Utility function)
    @staticmethod
    def is_valid_salary(salary: float) -> bool:
        """Utility method independent of instance or class state."""
        return salary >= 20000

    # 3. Property (Getter)
    @property
    def salary(self) -> float:
        """Getter for salary attribute."""
        return self._salary

    # Property Setter (Validation)
    @salary.setter
    def salary(self, new_salary: float) -> None:
        """Setter with validation logic."""
        if not Employee.is_valid_salary(new_salary):
            raise ValueError("Salary must be at least 20,000 INR.")
        self._salary = new_salary


# Testing Class Method (Alternative Constructor)
emp1 = Employee.from_string("Vetri-50000")
print(f"Employee Created: {emp1.name}, Salary: {emp1.salary}")

# Testing Static Method
print("Is 15000 valid?:", Employee.is_valid_salary(15000))
print("Is 30000 valid?:", Employee.is_valid_salary(30000))

# Testing Property Setter with Validation
emp1.salary = 60000  # Updated successfully
print("Updated Salary:", emp1.salary)

try:
    emp1.salary = 10000  # Should raise ValueError
except ValueError as e:
    print("Validation Error caught:", e)
```

# 03. Encapsulation and Access Modifiers

### Step 1: Understand the Concept
Encapsulation restricts direct access to an object's components, preventing unexpected modifications and exposing only necessary interfaces. Python enforces access controls using naming conventions:

* **Public Attributes (`var`)**: Accessible from anywhere inside or outside the class.
* **Protected Attributes (`_var`)**: Indicated by a single leading underscore. Intended for internal use within the class and its subclasses.
* **Private Attributes (`__var`)**: Indicated by double leading underscores. Triggers **Name Mangling** (`_ClassName__var`) to prevent accidental direct access or modification from outside the class.

### Step 2: Real-World Analogy
* **Public**: A company’s official website URL—anyone can view and share it.
* **Protected**: Internal company memo—intended only for employees and official partners.
* **Private**: A personal ATM PIN or vault password—strictly hidden from the outside world and accessible only through secured systems.

### Step 3: Implementation & Explanation

```python
class BankAccount:
    def __init__(self, account_holder: str, initial_balance: float, ssn_last4: str):
        self.account_holder = account_holder  # Public attribute
        self._account_type = "Savings"        # Protected attribute
        self.__balance = initial_balance      # Private attribute
        self.__ssn = ssn_last4                # Private attribute

    # Public Getter for Private Attribute
    def get_balance(self) -> float:
        return self.__balance

    # Public Setter with Business Validation
    def deposit(self, amount: float) -> None:
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount} INR. New Balance: {self.__balance} INR")
        else:
            print("Invalid deposit amount!")


# Instantiating BankAccount
account = BankAccount("Vetri", 5000.0, "4321")

# Accessing Public and Protected Attributes
print("Account Holder (Public):", account.account_holder)
print("Account Type (Protected):", account._account_type)

# Accessing Private Attribute via Public Method
print("Account Balance (via Getter):", account.get_balance())
account.deposit(1500.0)

# Attempting Direct Access to Private Attribute (Will raise AttributeError)
try:
    print(account.__balance)
except AttributeError as e:
    print("\nDirect Private Access Error:", e)

# Demonstrating Python Name Mangling
print("Private Access via Name Mangling:", account._BankAccount__balance)
```

**Code Explanation:**
* `self.account_holder` is public and directly accessible via `account.account_holder`.
* `self._account_type` uses a single underscore convention signaling protected scope for subclasses.
* `self.__balance` uses double underscores, triggering Python Name Mangling to transform the variable name internally to `_BankAccount__balance`.
* Direct access (`account.__balance`) raises an `AttributeError`, maintaining encapsulated integrity.
* Controlled modifications are safely routed through `deposit()` and `get_balance()`.

# 04. Abstraction (Abstract Base Classes - ABC)

### Step 1: Understand the Concept
Abstraction hides background implementation details and forces child classes to implement a specific contract/blueprint.
* **Abstract Base Class (ABC)**: A class that cannot be instantiated directly and serves as a formal template for subclasses.
* **`@abstractmethod`**: A decorator defining methods that **must** be overridden in any non-abstract subclass.

### Step 2: Real-World Analogy
* **Abstraction**: A payment gateway protocol. It defines that every payment method MUST implement `pay()`, but how a Credit Card vs. UPI vs. Net Banking processes the money is hidden inside their own individual implementations.

### Step 3: Implementation & Explanation

```python
from abc import ABC, abstractmethod


class PaymentGateway(ABC):

    @abstractmethod
    def process_payment(self, amount: float) -> None:
        """Abstract method forcing child classes to implement payment logic."""
        pass

    @abstractmethod
    def generate_receipt(self, transaction_id: str) -> str:
        """Abstract method forcing child classes to implement receipt generation."""
        pass

    def log_transaction(self, status: str) -> None:
        """Concrete method shared across all subclasses."""
        print(f"[LOG]: Payment status updated to: {status}")


class UPIPayment(PaymentGateway):

    def __init__(self, upi_id: str):
        self.upi_id = upi_id

    def process_payment(self, amount: float) -> None:
        print(f"Processing UPI payment of {amount} INR for {self.upi_id}")
        self.log_transaction("SUCCESS")

    def generate_receipt(self, transaction_id: str) -> str:
        return f"UPI-RECEIPT-{transaction_id}"


# 1. Attempting to instantiate an abstract class (Will raise TypeError)
try:
    gateway = PaymentGateway()
except TypeError as e:
    print("Cannot instantiate Abstract Class directly:", e)

# 2. Instantiating a concrete child class
upi = UPIPayment("harish@upi")
upi.process_payment(1500.0)
print("Receipt:", upi.generate_receipt("TXN9988"))
```

**Code Explanation:**
* `class PaymentGateway(ABC):` inherits from `ABC`, marking it as an abstract base class.
* `@abstractmethod` decorates `process_payment` and `generate_receipt`, preventing direct instantiation of `PaymentGateway`.
* `UPIPayment` implements all abstract methods, allowing successful object instantiation.

# 05. Inheritance Types and `super()`

### Step 1: Understand the Concept
Inheritance allows a child class to inherit attributes and methods from a parent class, promoting code reusability.
* **Single Inheritance**: One child inherits from one parent.
* **Multilevel Inheritance**: A child inherits from a parent, which itself inherits from a grandparent.
* **Multiple Inheritance**: A child inherits from multiple parent classes simultaneously.
* **Hierarchical Inheritance**: Multiple child classes inherit from a single parent class.
* **`super()`**: A built-in function used to call methods (including `__init__`) from a parent class.

### Step 2: Real-World Analogy
* **Inheritance**: Inheriting eye color or physical traits from a parent.
* **`super()`**: Asking your parent to pay the base bills before you add your own personal expenses on top.

### Step 3: Implementation & Explanation

```python
# 1. Base Class (Single Inheritance)
class Person:

    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def introduce(self) -> None:
        print(f"Hi, I am {self.name}, aged {self.age}.")


# Single Inheritance using super()
class Student(Person):

    def __init__(self, name: str, age: int, student_id: str):
        super().__init__(name, age)  # Reusing parent constructor
        self.student_id = student_id

    def introduce(self) -> None:
        super().introduce()
        print(f"Student ID: {self.student_id}")


# 2. Multiple Inheritance
class AcademicInfo:

    def __init__(self, department: str):
        self.department = department


class SportsInfo:

    def __init__(self, sport: str):
        self.sport = sport


class AthleteStudent(AcademicInfo, SportsInfo):

    def __init__(self, department: str, sport: str):
        AcademicInfo.__init__(self, department)
        SportsInfo.__init__(self, sport)


# Testing Single Inheritance
student1 = Student("Vetri", 21, "ST102")
student1.introduce()

# Testing Multiple Inheritance
athlete = AthleteStudent("Computer Science", "Cricket")
print(
    f"\nAthlete Dept: {athlete.department}, Sport Played: {athlete.sport}"
)
```

**Code Explanation:**
* `super().__init__(name, age)` delegates constructor calls to `Person`, avoiding duplicated initialization logic in `Student`.
* `super().introduce()` extends the parent method's behavior before adding student-specific details.
* `AthleteStudent` combines multiple parent classes (`AcademicInfo`, `SportsInfo`) into a single derived class.

# 06. Polymorphism (Method Overriding & Overloading)

### Step 1: Understand the Concept
Polymorphism allows objects of different classes to respond to the same method call in their own unique ways.
* **Method Overriding**: A child class provides its own custom implementation of a method that is already defined in its parent class.
* **Method Overloading (Python Approach)**: Unlike languages like C++ or Java, Python does not support traditional method overloading natively (same method name with different parameters). Instead, Python achieves overloading using default arguments (`default=None`), variable arguments (`*args`), or type checking.

### Step 2: Real-World Analogy
* **Polymorphism**: The "Play" button on a media player. Clicking "Play" behaves differently whether you are playing an MP3 audio file, a 4K video file, or an online stream, but the interface ("Play") remains identical.

### Step 3: Implementation & Explanation

```python
# 1. Method Overriding
class Shape:

    def area(self) -> float:
        return 0.0


class Rectangle(Shape):

    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    # Overriding parent's area() method
    def area(self) -> float:
        return self.width * self.height


class Circle(Shape):

    def __init__(self, radius: float):
        self.radius = radius

    # Overriding parent's area() method
    def area(self) -> float:
        return 3.14159 * (self.radius**2)


# 2. Method Overloading (Pythonic Approach using Default Args / *args)
class Calculator:

    # Overloading using optional arguments
    def add(
        self, a: float, b: float, c: float | None = None
    ) -> float:
        if c is not None:
            return a + b + c
        return a + b


# Demonstrating Method Overriding
shapes: list[Shape] = [Rectangle(5.0, 4.0), Circle(3.0)]

print("--- Method Overriding Output ---")
for shape in shapes:
    print(f"Shape Area: {shape.area()}")

# Demonstrating Method Overloading
calc = Calculator()
print("\n--- Method Overloading Output ---")
print("Sum of 2 numbers:", calc.add(10, 20))
print("Sum of 3 numbers:", calc.add(10, 20, 30))
```

**Code Explanation:**
* `Rectangle` and `Circle` both inherit from `Shape` and override the `area()` method to calculate their respective geometric areas dynamically.
* Polymorphic execution allows iterating through `shapes` and invoking `.area()` without needing to check the specific class type beforehand.
* `Calculator.add()` handles different parameter counts using optional default argument `c=None` to simulate method overloading in a Pythonic manner.

# 07. Object Relationships (Association, Aggregation, Composition)

### Step 1: Understand the Concept
Object-Oriented Design defines how objects interact and depend on each other:
* **Association**: A weak relationship where two objects interact independently (e.g., Student and Teacher). Neither owns the other.
* **Aggregation ("Has-A" / Weak Ownership)**: A relationship where the child object can exist independently outside the parent container (e.g., Department and Teacher). If the Department is deleted, the Teacher still exists.
* **Composition ("Part-Of" / Strong Ownership)**: A strong binding where child objects cannot exist without the parent (e.g., Car and Engine). If the Car is destroyed, the Engine is also destroyed.

### Step 2: Real-World Analogy
* **Aggregation**: A Library and Books. If the library building closes down, the books can be moved to another library.
* **Composition**: A Human Body and Brain. The brain cannot survive or function independently if the human body ceases to exist.

### Step 3: Implementation & Explanation

```python
# 1. Aggregation Example (Teacher exists independently of Department)
class Teacher:

    def __init__(self, name: str):
        self.name = name


class Department:

    def __init__(self, dept_name: str, teachers: list[Teacher]):
        self.dept_name = dept_name
        self.teachers = teachers  # Aggregation (passed from outside)


# 2. Composition Example (Engine is created INSIDE Car)
class Engine:

    def __init__(self, horsepower: int):
        self.horsepower = horsepower


class Car:

    def __init__(self, brand: str, horsepower: int):
        self.brand = brand
        self.engine = Engine(
            horsepower
        )  # Composition (tightly coupled inside)

    def display_specs(self) -> None:
        print(f"{self.brand} Car powered by {self.engine.horsepower} HP Engine")


# Testing Aggregation
t1 = Teacher("Dr. Vetri")
dept = Department("Computer Science", [t1])
print(
    f"Department: {dept.dept_name}, Teacher: {dept.teachers[0].name}"
)  # Teacher still exists independently as `t1`

# Testing Composition
my_car = Car("BMW", 400)
my_car.display_specs()
```

**Code Explanation:**
* `Department` receives a pre-created `Teacher` instance in its constructor (**Aggregation**). If `dept` is deleted, `t1` remains intact in memory.
* `Car` creates an `Engine` directly inside its `__init__` method (**Composition**). The lifetime of the `Engine` object is strictly tied to the `Car` object.**Code Explanation:**

# 08. Magic (Dunder) Methods (`__str__`, `__repr__`, `__len__`)

### Step 1: Understand the Concept
Magic or **Dunder** (Double Underscore) methods allow custom classes to emulate built-in Python behaviors like string representations, iteration, length evaluation, and arithmetic operations:
* **`__str__()`**: Defines a readable, user-friendly string representation (triggered by `print()` or `str()`).
* **`__repr__()`**: Defines an unambiguous, developer-focused representation used for debugging (triggered by interactive shells or `repr()`).
* **`__len__()`**: Defines custom behavior for the built-in `len()` function.

### Step 2: Real-World Analogy
* **`__str__`**: The polished cover/title of a book shown on a bookstore display shelf.
* **`__repr__`**: The ISBN code, printing press details, and barcode printed inside the inner cover for inventory management.

### Step 3: Implementation & Explanation

```python
class Book:

    def __init__(self, title: str, author: str, page_count: int):
        self.title = title
        self.author = author
        self.page_count = page_count

    # User-friendly string display
    def __str__(self) -> str:
        return f"'{self.title}' by {self.author}"

    # Developer debugging representation
    def __repr__(self) -> str:
        return f"Book(title='{self.title}', author='{self.author}', page_count={self.page_count})"

    # Overriding length behavior
    def __len__(self) -> int:
        return self.page_count


# Testing Dunder Methods
book1 = Book("Python OOP Mastery", "Vetri", 350)

# Triggers __str__()
print("User View (str):", str(book1))

# Triggers __repr__()
print("Developer View (repr):", repr(book1))

# Triggers __len__()
print("Book Length (pages):", len(book1))
```

**Code Explanation:**
* Overriding `__str__` customizes how `print(book1)` displays the object readable to end users.
* Overriding `__repr__` provides full structural state info for developer debugging.
* Overriding `__len__` maps Python's built-in `len(book1)` directly to the internal `page_count` attribute.

# 09. Bonus Topics: Dataclasses & Singleton Pattern

### Step 1: Understand the Concept
* **Dataclasses (`@dataclass`)**: A standard library decorator that automatically generates boilerplate methods like `__init__()`, `__repr__()`, and `__eq__()` for data-oriented classes.
* **Singleton Pattern**: A design pattern that restricts a class to instantiating only a single instance throughout the application lifecycle (e.g., Database connections or Global settings).

### Step 2: Real-World Analogy
* **Dataclass**: An automated form printer that automatically prints name tags and identity badges without filling each line manually.
* **Singleton**: The President of a country. There can only be one active President at any given time managing the government.

### Step 3: Implementation & Explanation

```python
from dataclasses import dataclass

# 1. Dataclass Example
@dataclass
class Product:
    name: str
    price: float
    quantity: int = 1

    def total_cost(self) -> float:
        return self.price * self.quantity


# 2. Singleton Design Pattern Example
class DatabaseConnection:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            print("Creating new Database Connection instance...")
            cls._instance = super().__new__(cls)
        return cls._instance


# Testing Dataclass
p1 = Product("Laptop", 55000.0, 2)
p2 = Product("Laptop", 55000.0, 2)

print("Dataclass Auto-generated String (__repr__):", p1)
print("Dataclass Total Cost:", p1.total_cost())
print("Automatic Equality Check (p1 == p2):", p1 == p2)

print("\n--- Singleton Pattern Test ---")
db1 = DatabaseConnection()
db2 = DatabaseConnection()

print("Are db1 and db2 the exact same instance in memory?:", db1 is db2)
```

**Code Explanation:**
* `@dataclass` automatically generates `__init__`, `__repr__`, and value-based `__eq__` comparisons, reducing repetitive boilerplate code.
* `DatabaseConnection` uses `__new__` to intercept instance creation, ensuring only a single object reference (`_instance`) exists in memory.
