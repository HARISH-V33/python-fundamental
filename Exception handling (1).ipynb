# 01. Exception Handling Foundations & Core Syntax

### Step 1: Understand the Concept
Exception handling prevents application crashes by intercepting runtime errors and routing execution into safe fallback routines.
* **`try` block**: Encloses code that might potentially raise an exception.
* **`except` block**: Catches and handles specific exceptions raised within the `try` block.
* **`else` block**: Executes **only** if no exceptions were raised inside the `try` block.
* **`finally` block**: Executes **unconditionally** regardless of whether an exception occurred (ideal for resource cleanup).
* **`raise`**: Manually triggers built-in or custom exceptions when business validation rules fail.

### Step 2: Real-World Analogy
* **`try`**: Ordering food online.
* **`except`**: Restaurant out of stock (handles the error gracefully).
* **`else`**: Order successful, food arrives (executes on success).
* **`finally`**: Cleaning the dining table (executes unconditionally regardless of whether you ate or not).

### Step 3: Implementation & Explanation

```python
# 1. Custom Exception Definition
class InsufficientFundsError(Exception):
    """Raised when withdrawal amount exceeds current account balance."""

    def __init__(self, balance: float, amount: float):
        self.balance = balance
        self.amount = amount
        super().__init__(
            f"Cannot withdraw {amount} INR. Current balance is only {balance} INR."
        )


# 2. Complete Flow Control Class (try, except, else, finally, raise)
class BankAccount:

    def __init__(self, balance: float):
        self.balance = balance

    def withdraw(self, amount: float) -> None:
        try:
            if amount > self.balance:
                # Raising Custom Exception
                raise InsufficientFundsError(self.balance, amount)
            self.balance -= amount
        except InsufficientFundsError as e:
            print(f"[EXCEPT BLOCK]: Withdrawal Failed -> {e}")
        else:
            print(
                f"[ELSE BLOCK]: Transaction Successful! Remaining Balance: {self.balance} INR"
            )
        finally:
            print(
                "[FINALLY BLOCK]: Audit log updated & session cleaned up."
            )


# Testing Execution
account = BankAccount(5000.0)

print("--- Test Case 1: Valid Withdrawal ---")
account.withdraw(2000.0)

print("\n--- Test Case 2: Invalid Withdrawal (Triggers Exception) ---")
account.withdraw(4000.0)
```

**Code Explanation:**
* `InsufficientFundsError` inherits from `Exception` to create a custom domain-specific error.
* `raise` explicitly halts normal execution in `withdraw()` when validation fails.
* `else` executes only on valid transactions when no exception occurs.
* `finally` runs unconditionally after every attempt to guarantee audit logging.

# 02. Assertions and 10 Standard Built-in Exceptions

### Step 1: Understand the Concept
* **`assert` Statement**: A debugging aid that evaluates a condition. If `False`, Python immediately raises an `AssertionError`.
* **Built-in Exceptions**: Standard internal exceptions built into Python to handle runtime anomalies.

### Step 2: Implementation & Explanation
Below is a demonstration of `assert` along with 10 standard built-in exceptions: how they occur and how to handle them cleanly.

```python
# 1. Demonstration of Assertions
def calculate_discount(price: float, discount_percent: float) -> float:
    assert price >= 0, "Price cannot be negative!"
    assert (
        0 <= discount_percent <= 100
    ), "Discount percentage must be between 0 and 100!"
    return price - (price * (discount_percent / 100))


try:
    print("Discounted Price:", calculate_discount(1000, 15))
    calculate_discount(1000, 150)  # Triggers AssertionError
except AssertionError as e:
    print("AssertionError Caught:", e)

print("\n" + "=" * 50 + "\n")


# 2. Demonstration of 10 Built-in Exceptions
def demonstrate_ten_builtin_exceptions():

    # 1. ValueError (Right data type, invalid value)
    try:
        num = int("invalid_integer")
    except ValueError as e:
        print(f"[1. ValueError Handled]: {e}")

    # 2. TypeError (Incompatible operation between types)
    try:
        res = "Text" + 10
    except TypeError as e:
        print(f"[2. TypeError Handled]: {e}")

    # 3. IndexError (Accessing list index out of bounds)
    try:
        items = [10, 20]
        val = items[5]
    except IndexError as e:
        print(f"[3. IndexError Handled]: {e}")

    # 4. KeyError (Accessing non-existent dictionary key)
    try:
        user = {"name": "Vetri"}
        age = user["age"]
    except KeyError as e:
        print(f"[4. KeyError Handled]: {e}")

    # 5. AttributeError (Calling non-existent method on an object)
    try:
        number = 100
        number.append(5)
    except AttributeError as e:
        print(f"[5. AttributeError Handled]: {e}")

    # 6. NameError (Referencing an undeclared variable)
    try:
        print(undefined_variable_name)
    except NameError as e:
        print(f"[6. NameError Handled]: {e}")

    # 7. FileNotFoundError (Accessing a file path that does not exist)
    try:
        with open("non_existent_file.txt", "r") as f:
            f.read()
    except FileNotFoundError as e:
        print(f"[7. FileNotFoundError Handled]: {e}")

    # 8. ZeroDivisionError (Dividing a number by zero)
    try:
        res = 50 / 0
    except ZeroDivisionError as e:
        print(f"[8. ZeroDivisionError Handled]: {e}")

    # 9. ImportError (Importing a non-existent attribute from a valid module)
    try:
        from math import non_existent_function
    except ImportError as e:
        print(f"[9. ImportError Handled]: {e}")

    # 10. ModuleNotFoundError (Importing a non-existent module file/package)
    try:
        import non_existent_module
    except ModuleNotFoundError as e:
        print(f"[10. ModuleNotFoundError Handled]: {e}")


demonstrate_ten_builtin_exceptions()
```

**Code Explanation & Exception Breakdown:**
1. **`ValueError`**: Raised when a function gets an argument of correct type but improper value (e.g., `int('abc')`).
2. **`TypeError`**: Occurs when an operation is applied to an inappropriate type (e.g., adding `str` and `int`).
3. **`IndexError`**: Occurs when accessing a sequence index out of its range.
4. **`KeyError`**: Raised when looking up a missing key in a dictionary.
5. **`AttributeError`**: Triggered when an attribute or method reference fails on an object.
6. **`NameError`**: Occurs when accessing a variable name that is not defined in the current local/global scope.
7. **`FileNotFoundError`**: Raised when attempting to open or read a non-existent file on disk.
8. **`ZeroDivisionError`**: Raised when division or modulo operation is performed with zero as denominator.
9. **`ImportError`**: Triggered when Python fails to import a specified name from a module.
10. **`ModuleNotFoundError`**: A subclass of `ImportError` raised when a specified module cannot be located.
