# 01. Python Basics
# 1. Variables
### Step 1: Understand the Concept
A variable is a named container in memory used to store data values. In Python, variables are dynamically typed, meaning you do not need to explicitly declare the data type—Python automatically infers it at runtime based on the assigned value.

### Step 2: Real-World Analogy
Think of a variable as a labeled storage box in a warehouse. The label on the box is the variable name, and whatever item you put inside the box is the value. You can swap the contents of the box anytime.

### Step 3: Implementation & Explanation


```python
# Creating variables with different data types
name = "Vetri"   # String variable
age = 21         # Integer variable
gpa = 8.5        # Float variable
is_enrolled = True   # Boolean variable

# Displaying variable values
print("Name:",name)
print("Age:",age)
print("GPA:",gpa)
print("Is Enrolled:", is_enrolled)
```

### Code Explanation:

. name,age, gpa, and is_enrolled are variables holding different data types.

. Python automatically assigns data types dynamically without requiring explicit type declarations like int or String.

# 2. Data Types
### Step 1: Understand the Concept
Data types specify the type of value a variable can hold. Python provides several built-in fundamental data types:

int: Whole numbers without decimals.

float: Numbers with floating-point decimal values.

str: Textual data enclosed in single or double quotes.

### Step 2: Real-World Analogy
Think of a job application form where different fields require specific input formats: Age requires whole numbers (int), height requires decimals (float), name requires characters (str), and an agreement checkbox is yes/no (bool).

### Code Explanation:

# 3. Type Casting
### Step 1: Understand the Concept
Type casting (or Type Conversion) is the process of converting a variable from one data type to another.

Implicit Conversion: Python automatically converts one data type to another without data loss (e.g., adding an integer to a float).

Explicit Conversion: The programmer manually converts data types using built-in functions like int(), float(), str().

### Step 2: Real-World Analogy
Think of currency exchange before traveling abroad. Converting USD to INR is like explicit type casting, where data/value is transformed into a usable format for a specific system.

### Step 3: Implementation & Explanation

```python
# 1. Implicit Type Conversion
num_int = 10
num_float = 5.5
result = num_int + num_float   # Automatically converted to float
print("Implicit Result:", result, "| Type:", type(result))

# 2. Explicit Type Conversion
str_number = "100"             # String representation of a number
converted_int = int(str_number) # Converting string to integer

print("Explicit Int:", converted_int, "| Type:", type(converted_int))
```

### Code Explanation:

Python implicitly converts num_int to a float during addition to preserve decimal precision.

int(str_number) explicitly converts the numeric string "100" into a valid integer 100 so mathematical operations can be performed on it.

# 4. Operators
### Step 1: Understand the Concept
Operators are special symbols used to carry out mathematical, logical, and relational evaluations on variables and values.

Arithmetic: +, -, *, /, % (Modulus), // (Floor Division).

Comparison: ==, !=, >, <, >=, <=.

Logical: and, or, not.

### Step 2: Real-World Analogy
Consider a supermarket billing counter. Addition (+) sums the item prices, comparison (>) checks if your order qualifies for free delivery, and logical and checks if your account balance is sufficient and your coupon code is valid.

### Code Explanation:

a // b calculates floor division, stripping away fractional decimal points.

(a > 10) and (b < 5) evaluates to True because both individual conditions are valid.

# 02. Control Flow
# 1. Conditional Statements (if, elif, else)
### Step 1: Understand the Concept
Conditional statements control the flow of execution based on boolean conditions.

if tests the primary condition.

elif (else if) tests subsequent conditions if the preceding ones evaluate to False.

else catches any remaining cases when all previous conditions fail.

### Step 2: Real-World Analogy
Think of a traffic light signal: If the light is Red, you stop; Else If the light is Yellow, you prepare to stop; Else (Green), you drive forward.

### Step 3: Implementation & Explanation

```python
# Program to categorize marks into academic grades
marks = 85

if marks >= 90:
    print("Grade: A+")
elif marks >= 75:
    print("Grade: A")
elif marks >= 50:
    print("Grade: B")
else:
    print("Grade: Fail")
```

### Code Explanation:

Python evaluates marks >= 90 first. Since it is False, it moves to marks >= 75.

Since 85 is greater than 75, it prints "Grade: A" and bypasses the remaining branches.

# 3. while Loop
### Step 1: Understand the Concept
A while loop continuously executes a block of code as long as a specified boolean condition remains True.

### Step 2: Real-World Analogy
Think of an alarm clock continuously ringing while the time matches your alarm time until you manually press the snooze or stop button.

### Step 3: Implementation & Explanation

```python
# Countdown timer implementation
count = 3

while count > 0:
    print(f"Countdown: {count}")
    count -= 1  # Decrementing the counter

print("Sprint execution started!")
```

# 4. Loop Control Statements (break, continue, pass)
### Step 1: Understand the Concept
Loop control statements alter the normal execution sequence of a loop:

break: Exits the loop immediately.

continue: Skips the remainder of the current iteration and jumps to the next one.

pass: Acts as a null placeholder where code is syntactically required but no action is needed.

### Step 2: Real-World Analogy
break: Leaving a billing queue immediately when you realize you forgot your wallet.

continue: Skipping a damaged item on a shelf and inspecting the next item.

pass: Leaving a placeholder note on a document for a section you plan to draft later.

### Step 3: Implementation & Explanation

```python
numbers = [1, 2, 3, 4, 5, 6]

print("--- Demonstrating Continue (Skip Even Numbers) ---")
for num in numbers:
    if num % 2 == 0:
        continue
    print(f"Odd Number: {num}")

print("\n--- Demonstrating Break (Stop at 4) ---")
for num in numbers:
    if num == 4:
        print("Reached 4! Terminating loop.")
        break
    print(f"Number: {num}")

print("\n--- Demonstrating Pass ---")
for num in numbers:
    if num == 3:
        pass  # Placeholder for future processing logic
    print(f"Processed: {num}")
```

### Code Explanation:

continue skips the print() statement whenever num is even.

break terminates the entire loop when num == 4.

pass serves as a syntactically valid empty block when num == 3.

# 04. Functions
### 1. Function Definition & Arguments
### Step 1: Understand the Concept
A function is a reusable block of code designed to perform a specific task. Functions prevent code duplication and promote modularity. Parameters are variables declared in the function definition, while arguments are the actual values passed during function invocation.

### Step 2: Real-World Analogy
Think of a juice blender machine. The machine is the function, the fruits you put in are the arguments, and the blended juice is the returned output.

### Step 3: Implementation & Explanation

```python
# Function definition with parameters
def calculate_total(price, quantity):
    total = price * quantity
    return total

# Function call with positional arguments
bill_amount = calculate_total(150, 3)
print("Total Bill Amount:", bill_amount)
```

### 2. Default and Keyword Arguments
### Step 1: Understand the Concept
Default Arguments: Fallback parameter values used if no explicit argument is passed during the function call.

Keyword Arguments: Passing arguments using name=value syntax, allowing arguments to be supplied in any order.

### Step 2: Real-World Analogy
Ordering coffee at a cafe: the default milk option might be standard dairy milk unless you explicitly specify oat milk via a keyword selection.

### Step 3: Implementation & Explanation

```python
# Function with default argument (role="Trainee")
def greet_employee(name, role="Trainee"):
    return f"Employee: {name} | Role: {role}"

# Using default argument
print(greet_employee("harish"))

# Overriding default argument using keyword argument
print(greet_employee(role="AI/ML Engineer", name="Stackly User"))
```

### Code Explanation:

role="Trainee" sets a default value.

Passing role="AI/ML Engineer"  explicitly overrides the default value regardless of argument position.

### 3. Variable-Length Arguments (*args and **kwargs)
### Step 1: Understand the Concept
*args: Allows a function to accept any number of positional arguments, collected as a tuple.

**kwargs: Allows a function to accept any number of keyword arguments, collected as a dictionary.

### Step 2: Real-World Analogy

### Code Explanation:

*args captures (10, 20, 30) as a tuple.

**kwargs captures key-value pairs as {'student': 'Harish', 'sprint': 'Sprint 1', 'status': 'Active'}.

### 4. Lambda Functions
### Step 1: Understand the Concept
A Lambda function is a small, anonymous function defined in a single line without using the def keyword. It can take any number of arguments but evaluates only a single expression.
### Step 2: Real-World Analogy
Think of a pocket calculator shortcut button—designed for a single quick calculation rather than a full multi-step formula sheet.

### Step 3: Implementation & Explanation


```python
# Lambda function for calculating square of a number
square = lambda x: x ** 2

# Lambda function for addition
add_numbers = lambda a, b: a + b

print("Square of 5:", square(5))
print("Sum of 10 and 20:", add_numbers(10, 20))
```

### Code Explanation:

### 5. Recursive Functions
### Step 1: Understand the Concept
A recursive function is a function that calls itself to solve a problem by breaking it down into smaller sub-problems. Every recursive function must have a base case to terminate execution and prevent stack overflow errors.

### Step 2: Real-World Analogy
Think of Russian nesting dolls (Matryoshka). You open one doll to find a smaller doll inside, repeating the action until you reach the smallest solid doll (base case) that cannot be opened further.

### Step 3: Implementation & Explanation

```python
# Recursive function to calculate factorial
def factorial(n):
    # Base Case
    if n == 1 or n == 0:
        return 1
    # Recursive Step
    return n * factorial(n - 1)

print("Factorial of 5:", factorial(5))
```

### Code Explanation:

factorial(5) executes as 5 * factorial(4), unwinding down to n == 1 (Base case) which returns 1 and stops further calls.

### 1. Iterators

### 2. Generators and yield
### Step 1: Understand the Concept
A Generator is a special type of function that returns an iterator, generating values lazily on-the-fly using the yield keyword instead of return. This makes generators highly memory-efficient when dealing with massive datasets.

### Step 2: Real-World Analogy
Think of video streaming (like YouTube). Instead of downloading an entire 4K movie into your RAM at once, the video streams in small chunks (yields) as you watch.

### Step 3: Implementation & Explanation

```python
# Generator function yielding numbers
def count_up_to(max_val):
    count = 1
    while count <= max_val:
        yield count
        count += 1

# Consuming generator
counter = count_up_to(3)
for num in counter:
    print("Generated Number:", num)
```

### Code Explanation:

The yield keyword pauses function execution, saves its state, and returns the current value, resuming from the same spot on the next request.

# 3. Decorators
### Step 1: Understand the Concept
A Decorator is a function that takes another function as an argument, extends or modifies its behavior without altering its source code, and returns the enhanced function.

### Step 2: Real-World Analogy
Think of putting a protective mobile phone case on your phone. The case adds extra protection and features (like a kickstand) without altering the internal circuitry of the phone itself.

### Step 3: Implementation & Explanation

```python
# Decorator Function
def log_execution(func):
    def wrapper():
        print("[LOG] Function execution started...")
        func()
        print("[LOG] Function execution finished.")
    return wrapper

# Applying decorator using @ syntax
@log_execution
def greet():
    print("Hello! Welcome to Sprint 1.")

# Function Call
greet()
```

### Code Explanation:

@log_execution wraps greet(), executing wrapper code before and after calling the core greet() logic dynamically.

# 4. Closures
### Step 1: Understand the Concept
A Closure is an inner function that retains access to variables from its outer enclosing scope even after the outer function has finished executing.

### Step 2: Real-World Analogy
Think of a backpack packed with snacks before leaving home. Even after you leave the house (outer function terminates), you still carry access to the packed snacks (inner function access).

### Step 3: Implementation & Explanation


```python
# Outer function
def multiplier(factor):
    # Inner function forms a closure
    def multiply_by(number):
        return number * factor
    return multiply_by

# Creating closure instances
double = multiplier(2)
triple = multiplier(3)

print("Double 5:", double(5)) # Output: 10
print("Triple 5:", triple(5)) # Output: 15
```

### Code Explanation:

double remembers factor = 2 even after multiplier() has finished running, demonstrating closure state retention.

# 5. global and nonlocal Keywords
### Step 1: Understand the Concept
global: Used to modify a variable declared at the module/top-level scope from within a function.

nonlocal: Used inside nested functions to modify a variable declared in the nearest outer (non-global) function scope.

### Step 2: Real-World Analogy
global: Changing the main street lighting schedule affecting the entire town.

nonlocal: Adjusting the air conditioner setting inside a specific room, affecting the apartment scope without altering the city grid.

### Step 3: Implementation & Explanation

```python
# Demonstrating global keyword
counter = 0

def increment_global():
    global counter
    counter += 1

increment_global()
print("Global Counter:", counter)

# Demonstrating nonlocal keyword
def outer_func():
    message = "Outer"
    def inner_func():
        nonlocal message
        message = "Modified Outer"
    inner_func()
    print("Outer Message:", message)

outer_func()
```

### Code Explanation:

global counter allows modifying the module-level variable inside increment_global().

nonlocal message allows inner_func() to update the enclosing scope variable defined in outer_func().

## 2. Type Hinting

### Step 1: Understand the Concept
Type Hinting is a formal syntax introduced in Python 3.5 that allows developers to specify the expected data types for function arguments, return values, and variables. It enhances code readability and enables static type checking tools (like `mypy`).

### Step 2: Real-World Analogy
Think of labeled slots in a coin sorting machine. The machine labels slots for 1, 5, and 10 Rupee coins so you instantly know what goes where without guessing.

```python
# Function with Type Hints
def calculate_bmi(weight_kg: float, height_m: float) -> float:
    bmi: float = weight_kg / (height_m ** 2)
    return bmi

# Function Invocation
user_bmi = calculate_bmi(70.5, 1.75)
print(f"Calculated BMI: {user_bmi:.2f}")
```

**Code Explanation:**
* `weight_kg: float` specifies that the input argument should ideally be a floating-point number.
* `-> float` indicates that the function will return a float value.
* Type hints do not crash execution at runtime if mismatched, but serve as developer documentation.

# 07. Built-in Functions (`map`, `filter`, `zip`, `enumerate`, `sorted`)

### Step 1: Understand the Concept
Python provides high-performance functional built-in utilities:
* **`map()`**: Transforms all items in an iterable using a specified function.
* **`filter()`**: Filters elements from an iterable based on a boolean condition function.
* **`zip()`**: Aggregates elements from two or more iterables into tuple pairs.
* **`enumerate()`**: Returns an iterator yielding tuple pairs containing index counters along with values.
* **`sorted()`**: Returns a new sorted list from elements in any iterable.

### Step 2: Real-World Analogy
* **`map`**: Spray painting every car in an assembly line blue.
* **`filter`**: A security scanner allowing only passengers with valid tickets through.
* **`zip`**: Pairing socks together from a basket into matched pairs.
* **`enumerate`**: Assigning token queue numbers to visitors as they arrive.

### Step 3: Implementation & Explanation

```python
numbers = [1, 2, 3, 4, 5, 6]
names = ["Vetri", "Arun", "Karthik"]
scores = [85, 92, 78]

# 1. map(): Doubling all numbers
doubled = list(map(lambda x: x * 2, numbers))
print("Mapped (Doubled):", doubled)

# 2. filter(): Extracting numbers greater than 3
filtered = list(filter(lambda x: x > 3, numbers))
print("Filtered (> 3):", filtered)

# 3. zip(): Pairing names with their respective scores
paired_data = list(zip(names, scores))
print("Zipped Pairs:", paired_data)

# 4. enumerate(): Adding index tracking to a list
print("\nEnumerated Names:")
for index, name in enumerate(names, start=1):
    print(f"Rank {index}: {name}")

# 5. sorted(): Sorting scores in descending order
sorted_scores = sorted(scores, reverse=True)
print("\nSorted Scores (Descending):", sorted_scores)
```

**Code Explanation:**
* `map(lambda x: x * 2, numbers)` applies multiplication to each element in `numbers`.
* `filter(lambda x: x > 3, numbers)` retains only elements satisfying `x > 3`.
* `zip(names, scores)` pairs element-by-element: `("Vetri", 85)`, `("Arun", 92)`.
* `enumerate(names, start=1)` returns tuples like `(1, "Vetri")`, tracking iterations cleanly.
* `sorted(scores, reverse=True)` orders numeric elements from highest to lowest.
